```python
#Q2

def isPrime(n):
    if n < 2 : 
        return False
    
    correct=False
    for i in range(2,n):
        if(n%i == 0):
            correct=True
            break
        
    if(correct):
        return False
    else:
        return True
isPrime(14)
```

```python
# Q19: Load the CSV, check missing values, and filling numeric missing values with column mean

import pandas as pd

df1 = pd.read_csv("uncleaned_students_dataset.csv")

print("Missing values before filling:")
print(df1.isna().sum())

numeric_columns = df1.select_dtypes(include="number").columns

for column in numeric_columns:
    df1[column] = df1[column].fillna(df1[column].mean())

print("\nMissing values after filling:")
print(df1.isna().sum())

print("\nUpdated DataFrame:")
print(df1)
```
